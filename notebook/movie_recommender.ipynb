# DATA CLEANING AND PREPROCESSING

```python
# Creating copies of our original datasets

movies_clean = movies_df.copy()
credits_clean = credits_df.copy()
```

```python
print("Duplicate IDs in Movies:",
      movies_clean["id"].duplicated().sum())

print("Duplicate IDs in Credits:",
      credits_clean["movie_id"].duplicated().sum())
```

```python
movies_clean = movies_clean.drop_duplicates(
    subset="id",
    keep="first"
)

credits_clean = credits_clean.drop_duplicates(
    subset="movie_id",
    keep="first"
)

print("Duplicate records handled successfully!")
```

```python
# Combining both datasets using movie IDs

merged_df = movies_clean.merge(
    credits_clean,
    left_on="id",
    right_on="movie_id",
    how="left",
    suffixes=("", "_credits"),
    validate="one_to_one"
)

print("Datasets combined successfully!")

print("Combined Dataset Shape:", merged_df.shape)
```

```python
merged_df.head()
```

```python
print("Missing values in Combined Dataset:")
print(merged_df.isnull().sum().sort_values(ascending=False).head(15))
```

```python
# Filling missing text information with empty strings

merged_df["overview"] = merged_df["overview"].fillna("")

# Filling missing movie titles and removing empty titles
merged_df["title"] = merged_df["title"].fillna("").astype(str).str.strip()

merged_df = merged_df[merged_df["title"] != ""].copy()

print("Missing values in important text columns handled!")
```

```python
def parse_list(value):

    if not isinstance(value, str):
        return []

    try:
        result = ast.literal_eval(value)

        if isinstance(result, list):
            return result

        return []

    except (ValueError, SyntaxError):
        return []
```

```python
merged_df["genres_list"] = merged_df["genres"].apply(parse_list)

merged_df["keywords_list"] = merged_df["keywords"].apply(parse_list)

merged_df["cast_list"] = merged_df["cast"].apply(parse_list)

merged_df["crew_list"] = merged_df["crew"].apply(parse_list)

print("Converted genres, keywords, cast and crew successfully!")
```

```python
def extract_names(items, limit=None):

    names = []

    for item in items:

        if isinstance(item, dict):

            name = item.get("name")

            if name:
                names.append(name)

        if limit is not None and len(names) >= limit:
            break

    return names
```

```python
merged_df["genres_text"] = merged_df["genres_list"].apply(extract_names)

merged_df["keywords_text"] = merged_df["keywords_list"].apply(extract_names)

print("Genres and keywords extracted successfully!")
```

```python
merged_df["cast_text"] = merged_df["cast_list"].apply(
    lambda actors: extract_names(actors, limit=3)
)

print("Top three actors extracted successfully!")
```

```python
def get_director(crew):

    for person in crew:

        if isinstance(person, dict):

            if person.get("job") == "Director":

                return person.get("name", "")

    return ""
```

```python
merged_df["director"] = merged_df["crew_list"].apply(get_director)

print("Movie directors extracted successfully!")
```

```python
merged_df["genres_text"] = merged_df["genres_text"].apply(
    lambda x: " ".join(x)
)

merged_df["keywords_text"] = merged_df["keywords_text"].apply(
    lambda x: " ".join(x)
)

merged_df["cast_text"] = merged_df["cast_text"].apply(
    lambda x: " ".join(x)
)
```

```python
merged_df["tags"] = (
    merged_df["overview"] + " " +
    merged_df["genres_text"] + " " +
    merged_df["keywords_text"] + " " +
    merged_df["cast_text"] + " " +
    merged_df["director"]
)
```

```python
merged_df["tags"] = (
    merged_df["tags"]
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

print("Combined movie tags created successfully!")
```

```python
final_columns = [
    "id",
    "title",
    "overview",
    "genres_text",
    "keywords_text",
    "cast_text",
    "director",
    "tags"
]

# Include some additional information for displaying recommendations
optional_columns = [
    "release_date",
    "vote_average",
    "vote_count",
    "popularity"
]

for column in optional_columns:

    if column in merged_df.columns:
        final_columns.append(column)


recommendation_df = merged_df[final_columns].copy()

print("Final cleaned dataset created!")
print("Dataset Shape:", recommendation_df.shape)
```

```python
classification_df = merged_df.copy()

classification_df = classification_df[
    [
        "id",
        "title",
        "tags",
        "popularity",
        "runtime",
        "release_date",
        "original_language"
    ]
].copy()

print("Classification dataset prepared successfully!")

print("Dataset Shape:", classification_df.shape)

classification_df.head()
```

```python
classification_df["popularity"] = pd.to_numeric(
    classification_df["popularity"],
    errors="coerce"
)

classification_df["runtime"] = pd.to_numeric(
    classification_df["runtime"],
    errors="coerce"
)

classification_df["release_year"] = pd.to_datetime(
    classification_df["release_date"],
    errors="coerce"
).dt.year

classification_df["tags"] = classification_df["tags"].fillna("")

classification_df["original_language"] = (
    classification_df["original_language"].fillna("Unknown")
)

classification_df["runtime"] = (
    classification_df["runtime"].replace(0, np.nan)
)


classification_df = classification_df.dropna(
    subset=["popularity"]
).copy()

print("Classification data cleaned successfully!")
```

## 8.1 Building the Preprocessing Pipeline

```python
from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.impute import SimpleImputer

from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
numeric_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

print("Numerical preprocessing pipeline created!")
```

```python
categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(handle_unknown="ignore")
        )
    ]
)

print("Categorical preprocessing pipeline created!")
```

```python
text_pipeline = TfidfVectorizer(
    max_features=10000,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=2
)

print("Text preprocessing pipeline created!")
```

```python
preprocessor = ColumnTransformer(

    transformers=[

        (
            "text",
            text_pipeline,
            "tags"
        ),

        (
            "categorical",
            categorical_pipeline,
            ["original_language"]
        ),

        (
            "numerical",
            numeric_pipeline,
            ["runtime", "release_year"]
        )

    ]

)

print("Complete preprocessing pipeline created successfully!")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Recommendation libraries imported successfully!")
```

```python
# Load the cleaned movie dataset
recommendation_df = pd.read_csv(data_folder / "cleaned_movies.csv")

# Check the dataset
print("Dataset loaded successfully!")
print("Total movies:", len(recommendation_df))

display(recommendation_df.head())
```

```python
# Create TF-IDF vectorizer
tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words="english"
)

# Convert movie tags into numerical vectors
tfidf_matrix = tfidf.fit_transform(
    recommendation_df["tags"].fillna("")
)

print("TF-IDF applied successfully!")
print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

```python
def recommend_movies(movie_title, number_of_recommendations=10):

    # Find the selected movie
    matching_movies = recommendation_df[
        recommendation_df["title"].str.lower() == movie_title.lower()
    ]

    # Check whether the movie exists
    if matching_movies.empty:
        print("Movie not found. Please check the movie title.")
        return

    # Get the selected movie's index
    movie_index = matching_movies.index[0]

    # Calculate similarity with all other movies
    similarity_scores = cosine_similarity(
        tfidf_matrix[movie_index],
        tfidf_matrix
    ).flatten()

    # Sort movies by similarity score (highest first)
    similar_indices = similarity_scores.argsort()[::-1]

    # Collect the recommended movies
    recommended_indices = []

    for index in similar_indices:

        # Do not recommend the selected movie itself
        if index != movie_index:
            recommended_indices.append(index)

        if len(recommended_indices) == number_of_recommendations:
            break

    # Display recommendations
    recommendations = recommendation_df.iloc[recommended_indices]

    return recommendations[
        ["title", "genres_text", "director"]
    ].reset_index(drop=True)
```
